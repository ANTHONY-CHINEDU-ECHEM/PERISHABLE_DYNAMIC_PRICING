# EDA — Perishable Pricing Dataset

```python
import pandas as pd
df = pd.read_csv('../data/price_demand_history.csv')
df.groupby('category')['units_sold'].describe()
```

## Price-demand relationship by category

```python
import matplotlib.pyplot as plt
for cat, grp in df.groupby('category'):
    plt.scatter(grp['markdown_pct'], grp['units_sold'], alpha=0.1, s=5, label=cat)
plt.legend(); plt.xlabel('markdown %'); plt.ylabel('units sold'); plt.show()
```
